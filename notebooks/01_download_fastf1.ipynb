### Utility Functions

The following functions are responsible for:

- Creating session directories
- Saving datasets
- Writing metadata
- Logging download status

These functions are designed to be reusable throughout the project.

```python
from pathlib import Path
import json
from datetime import datetime

import pandas as pd
```

```python
# session directory creator

def create_session_directory(base_dir: Path,
                             season: int,
                             event_name: str,
                             session_name: str) -> Path:
    """
    Create directory structure for a session.

    Example:
    data/raw/fastf1/2018/Australian_Grand_Prix/Race/
    """

    safe_event = event_name.replace(" ", "_")
    safe_session = session_name.replace(" ", "_")

    session_path = (
        base_dir
        / str(season)
        / safe_event
        / safe_session
    )

    session_path.mkdir(parents=True, exist_ok=True)

    return session_path
```

```python
# save DataFrame

def save_dataframe(df: pd.DataFrame,
                   save_path: Path,
                   filename: str):
    """
    Save dataframe as parquet.
    """

    filepath = save_path / f"{filename}.parquet"

    df.to_parquet(filepath, index=False)

    print(f"✓ Saved : {filepath.name}")
```

```python
# ============================================================
# Metadata
# ============================================================

def save_metadata(session, save_path: Path):

    metadata = {

        "season": session.event["EventDate"].year,

        "round": int(session.event["RoundNumber"]),

        "event": session.event["EventName"],

        "official_name": session.event["OfficialEventName"],

        "country": session.event["Country"],

        "location": session.event["Location"],

        "session": session.name,

        "event_format": session.event["EventFormat"],

        "f1_api_support": bool(session.f1_api_support),

        "drivers": len(session.drivers),

        "total_laps": (
            int(session.total_laps)
            if session.total_laps is not None
            else None
        ),

        "download_time": datetime.now().isoformat(),

        "fastf1_version": fastf1.__version__

    }

    with open(
        save_path / "metadata.json",
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            metadata,
            file,
            indent=4,
            ensure_ascii=False
        )

    print("Metadata Saved")
```

```python
# Download Log

def save_download_log(save_path: Path,
                      status: dict):

    with open(save_path / "download_log.json",
              "w",
              encoding="utf-8") as file:

        json.dump(
            status,
            file,
            indent=4
        )

    print("Download Log Saved")
```

```python
# Driver Number → Driver Code Mapping


def get_driver_mapping(session):
    """
    Create a mapping from driver number to FIA driver code.

    Example:
    {
        '1': 'VER',
        '11': 'PER',
        '16': 'LEC'
    }
    """

    mapping = {}

    results = session.results

    for _, row in results.iterrows():

        number = str(row["DriverNumber"])
        code = row["Abbreviation"]

        mapping[number] = code

    return mapping
```

#### Generic Dataset Saving

Instead of writing separate save functions for every dataset, we use one
generic function that:

- Checks whether the dataset exists.
- Saves it if available.
- Logs success or failure.
- Prevents the download pipeline from stopping.

```python
# ============================================================
# Safe Session Dataset Saver
# ============================================================

def save_session_dataset(
    session,
    attribute_name: str,
    file_name: str,
    session_path: Path,
    download_status: dict
):
    """
    Safely access a Session DataFrame attribute and save it as parquet.
    """

    try:

        dataset = getattr(session, attribute_name)

        if dataset is None or len(dataset) == 0:

            download_status[file_name] = {
                "status": "Not Available",
                "rows": 0
            }

            print(f"{file_name} : Not Available")
            return

        dataset.to_parquet(
            session_path / f"{file_name}.parquet",
            index=False
        )

        download_status[file_name] = {
            "status": "Success",
            "rows": len(dataset)
        }

        print(f"Saved : {file_name} ({len(dataset)} rows)")

    except Exception as e:

        download_status[file_name] = {
            "status": "Failed",
            "reason": type(e).__name__
        }

        print(f"{file_name} : {type(e).__name__}")
```

#### Session Downloader

This function downloads one complete session and saves every available
dataset independently.

```python

# Download One Session

def download_session(
    session,
    session_path=None
                        ):
    if session_path is None:
        session_path = create_session_directory(
            FASTF1_DIR,
            session.event["EventDate"].year,
            session.event["EventName"],
            session.name
        )

    print(f"\nSaving to:\n{session_path}\n")

    download_status = {}

    # Save Standard Datasets


    save_session_dataset(
        session,
        "laps",
        "laps",
        session_path,
        download_status
    )

    save_session_dataset(
        session,
        "results",
        "results",
        session_path,
        download_status
    )

    save_session_dataset(
        session,
        "weather_data",
        "weather",
        session_path,
        download_status
    )

    try:

        save_telemetry_dict(
            session,
            session.car_data,
            "car_data",
            session_path,
            download_status
        )
    except Exception as e:

        download_status["car_data"] = f"failed : {e}"
        print("car_data unavailable")

    try:

        save_telemetry_dict(
            session,
            session.pos_data,
            "position_data",
            session_path,
            download_status
        )

    except Exception as e:

        download_status["position_data"] = f"failed:{e}"
        print("position_data_unavailable")

        


    # Metadata
    

    save_metadata(
        session,
        session_path
    )

    save_download_log(
        session_path,
        download_status
    )

    print("\nDownload Complete.\n")

    return download_status
```

```python
# ============================================================
# Save Telemetry Dictionary
# ============================================================

def save_telemetry_dict(
    session,
    telemetry_dict: dict,
    dataset_name: str,
    session_path: Path,
    download_status: dict
):
    """
    Save telemetry data for each driver as a separate parquet file.
    """

    try:

        folder = session_path / dataset_name
        folder.mkdir(exist_ok=True)

        driver_map = get_driver_mapping(session)

        saved = 0
        total_rows = 0

        for driver_number, telemetry in telemetry_dict.items():

            driver_code = driver_map.get(driver_number, driver_number)

            telemetry.to_parquet(
                folder / f"{driver_code}.parquet",
                index=False
            )

            saved += 1
            total_rows += len(telemetry)

        download_status[dataset_name] = {
            "status": "Success",
            "drivers": saved,
            "rows": total_rows
        }

        print(
            f"Saved {saved} drivers "
            f"({total_rows:,} rows)"
        )

    except Exception as e:

        download_status[dataset_name] = {
            "status": "Failed",
            "reason": type(e).__name__
        }

        print(f"{dataset_name} : {type(e).__name__}")
```

### Bulk Downloader

Download all Formula 1 sessions season-wise using FastF1.

The downloader automatically:

- Retrieves the official event schedule
- Downloads every available session
- Saves datasets
- Continues even if one session fails

```python

# Bulk Download Configuration


START_SEASON = 2022
END_SEASON = 2025

DOWNLOAD_SESSIONS = [

    "Practice 1",
    "Practice 2",
    "Practice 3",

    "Sprint Qualifying",
    "Sprint",

    "Qualifying",

    "Race"

]
```

```python

# Download Entire Season 

import time
from fastf1.req import RateLimitExceededError

def download_season(season):

    print("=" * 70)
    print(f"Season {season}")
    print("=" * 70)

    
    # Get Schedule (Retry on RateLimit)
    

    while True:

        try:

            schedule = fastf1.get_event_schedule(season)
            total_events = len(schedule)
            current_event = 0
            break

        except RateLimitExceededError:

            print("\nRate limit reached while fetching schedule.")
            print("Sleeping for 1 hour...\n")

            time.sleep(3600)

    
    # Events
    

    for _, event in schedule.iterrows():

        event_name = event["EventName"]

        print(
            f"\n[{current_event}/{total_events}] "
            f"{event_name}"
            )

        
        # Sessions
        

        for session_name in DOWNLOAD_SESSIONS:

            while True:

                try:

                    # Create session object
                    session = fastf1.get_session(
                        season,
                        event_name,
                        session_name
                    )

                    # Skip if already downloaded
                    session_path = create_session_directory(
                        FASTF1_DIR,
                        session.event["EventDate"].year,
                        session.event["EventName"],
                        session.name
                        )

                    if (session_path / "metadata.json").exists():

                        print(f"✓ Already Downloaded : {session_name}")
                        break

                    # Load data
                    session.load()

                    # Save everything
                    download_session(
                                    session,
                                    session_path
                                    )

                    # Session completed
                    break

                except RateLimitExceededError:

                    print("\nRate limit reached.")
                    print("Sleeping for 1 hour...\n")

                    time.sleep(3600)

                    # Retry SAME session
                    continue

                except Exception as e:

                    print(f"\n❌ {session_name}")
                    print(type(e).__name__)
                    print(e)

                    break
        current_event += 1

    print("\nSeason Finished\n")
```

```python
for season in range(
    START_SEASON,
    END_SEASON + 1
):

    download_season(season)
```

```python
#SUmmary Check

from pathlib import Path

def get_directory_size(path: Path):
    total = 0
    for f in path.rglob("*"):
        if f.is_file():
            total += f.stat().st_size
    return total / (1024**3)   # GB


season_dirs = sorted([d for d in FASTF1_DIR.iterdir() if d.is_dir() and d.name.isdigit()])

gp_count = 0
session_count = 0
parquet_count = 0
metadata_count = 0

for season in season_dirs:

    gps = [d for d in season.iterdir() if d.is_dir()]
    gp_count += len(gps)

    for gp in gps:

        sessions = [d for d in gp.iterdir() if d.is_dir()]
        session_count += len(sessions)

        for session in sessions:

            parquet_count += len(list(session.glob("*.parquet")))

            if (session / "metadata.json").exists():
                metadata_count += 1

size_gb = get_directory_size(FASTF1_DIR)

print("=" * 65)
print("            FASTF1 DATASET SUMMARY")
print("=" * 65)

print(f"Dataset            : FastF1")
print(f"Seasons            : {season_dirs[0].name} - {season_dirs[-1].name}")
print(f"Grand Prix         : {gp_count}")
print(f"Sessions           : {session_count}")
print(f"Parquet Files      : {parquet_count}")
print(f"Metadata Files     : {metadata_count}")
print(f"Dataset Size (GB)  : {size_gb:.2f}")

print("=" * 65)
```

```python
# Integrity Check
from pathlib import Path

missing_metadata = []
empty_sessions = []

for season in sorted([d for d in FASTF1_DIR.iterdir() if d.is_dir() and d.name.isdigit()]):

    for gp in season.iterdir():

        if not gp.is_dir():
            continue

        for session in gp.iterdir():

            if not session.is_dir():
                continue

            metadata = session / "metadata.json"
            parquet_files = list(session.glob("*.parquet"))

            if not metadata.exists():
                missing_metadata.append(session)

            if len(parquet_files) == 0:
                empty_sessions.append(session)

print("=" * 65)
print("              DATASET INTEGRITY CHECK")
print("=" * 65)

print(f"Missing Metadata : {len(missing_metadata)}")
print(f"Empty Sessions   : {len(empty_sessions)}")

if len(missing_metadata) == 0 and len(empty_sessions) == 0:
    print("\nSTATUS : PASSED ✅")
else:
    print("\nSTATUS : WARNING ⚠️")

print("=" * 65)
```
